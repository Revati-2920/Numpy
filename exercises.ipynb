Valid Sudoku

```python
import numpy as np
```

```python
s = np.random.randint(1,10,81).reshape(9,9)
s
```

```python
b = np.sum(s,axis=1)
for i in b:
    if i != 45:
        print("Sudoku is Not valid for rows")
        break
else:
    print("Sudoku is valid for rows")
```

```python
c = np.sum(s,axis=0)
for i in c:
    if i != 45:
        print("Sudoku is Not valid for columns")
        break
else:
    print("Sudoku is valid for rows")
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
        n = s[i:i+3,j:j+3]
        sum = n.sum()
        if sum != 45:
            print("Invalid Sudoku")
            break
    else:
        print("valid")
```
