```python
import numpy as np

# ============================================================
# STUDENT PERFORMANCE ANALYZER
# ============================================================

# Student names
students = np.array([
    "Aarav",
    "Priya",
    "Rahul",
    "Sneha",
    "Aditya",
    "Neha",
    "Rohan",
    "Isha"
])

# Subjects
subjects = np.array([
    "Math",
    "Physics",
    "Python",
    "DSA",
    "DBMS"
])

# Marks
marks = np.array([
    [85, 78, 92, 88, 81],
    [72, 65, 70, 75, 68],
    [95, 91, 89, 94, 96],
    [60, 72, 68, 65, 70],
    [88, 84, 90, 86, 85],
    [76, 80, 74, 79, 82],
    [91, 87, 93, 90, 88],
    [55, 62, 58, 64, 60]
])


# ============================================================
# TASK 1: Dataset Information
# ============================================================

print("=" * 50)
print("      STUDENT PERFORMANCE ANALYZER")
print("=" * 50)

num_students = marks.shape[0]
num_subjects = marks.shape[1]

print("\n--- DATASET INFORMATION ---")
print("Number of Students :", num_students)
print("Number of Subjects :", num_subjects)
print("Shape              :", marks.shape)
print("Dimensions         :", marks.ndim)
print("Total Marks Entries:", marks.size)


# ============================================================
# TASK 2: Total Marks of Each Student
# ============================================================

total_marks = np.sum(marks, axis=1)

print("\n--- TOTAL MARKS ---")

for i in range(num_students):
    print(f"{students[i]:8} -> {total_marks[i]}")


# ============================================================
# TASK 3: Average Marks of Each Student
# ============================================================

average_marks = np.mean(marks, axis=1)

print("\n--- AVERAGE MARKS ---")

for i in range(num_students):
    print(f"{students[i]:8} -> {average_marks[i]:.2f}")


# ============================================================
# TASK 4: Subject-wise Average
# ============================================================

subject_average = np.mean(marks, axis=0)

print("\n--- SUBJECT-WISE AVERAGE ---")

for i in range(num_subjects):
    print(f"{subjects[i]:8} : {subject_average[i]:.2f}")


# ============================================================
# TASK 5: Top and Lowest Performer
# ============================================================

top_index = np.argmax(average_marks)
lowest_index = np.argmin(average_marks)

print("\n--- TOP PERFORMER ---")
print("Name    :", students[top_index])
print("Average :", f"{average_marks[top_index]:.2f}")

print("\n--- LOWEST PERFORMER ---")
print("Name    :", students[lowest_index])
print("Average :", f"{average_marks[lowest_index]:.2f}")


# ============================================================
# TASK 6: Highest and Lowest Marks
# ============================================================

highest_mark = np.max(marks)
lowest_mark = np.min(marks)

print("\n--- HIGHEST / LOWEST MARK ---")
print("Highest Mark :", highest_mark)
print("Lowest Mark  :", lowest_mark)


# Bonus: Student and subject having the highest mark

highest_position = np.unravel_index(
    np.argmax(marks),
    marks.shape
)

highest_student_index = highest_position[0]
highest_subject_index = highest_position[1]

print("\n--- HIGHEST MARK DETAILS ---")
print("Student :", students[highest_student_index])
print("Subject :", subjects[highest_subject_index])
print("Mark    :", highest_mark)


# ============================================================
# TASK 7: PASS / FAIL
# ============================================================

# Student passes only if every subject mark >= 40
pass_status = np.all(marks >= 40, axis=1)

print("\n--- PASS / FAIL STATUS ---")

for i in range(num_students):

    if pass_status[i]:
        status = "PASS"
    else:
        status = "FAIL"

    print(f"{students[i]:8} -> {status}")


# ============================================================
# TASK 8: Students with Average Above 80
# ============================================================

above_80 = average_marks > 80

print("\n--- STUDENTS WITH AVERAGE ABOVE 80 ---")

for i in range(num_students):
    if above_80[i]:
        print(students[i])


# ============================================================
# TASK 9: Student Ranking
# ============================================================

# argsort gives indices in ascending order.
# [::-1] reverses them to descending order.

ranking_indices = np.argsort(average_marks)[::-1]

print("\n--- STUDENT RANKING ---")

for rank, index in enumerate(ranking_indices, start=1):
    print(
        f"{rank}. {students[index]:8} "
        f"Average: {average_marks[index]:.2f}"
    )


# ============================================================
# TASK 10: CLASS STATISTICS
# ============================================================

class_average = np.mean(marks)
median_marks = np.median(marks)
standard_deviation = np.std(marks)

best_subject_index = np.argmax(subject_average)

print("\n--- CLASS STATISTICS ---")

print("Class Average      :", f"{class_average:.2f}")
print("Median             :", f"{median_marks:.2f}")
print("Highest Mark       :", highest_mark)
print("Lowest Mark        :", lowest_mark)
print("Standard Deviation :", f"{standard_deviation:.2f}")
print("Best Subject       :", subjects[best_subject_index])
print("Best Subject Avg   :", f"{subject_average[best_subject_index]:.2f}")


# ============================================================
# BONUS 1: Best DSA Student
# ============================================================

dsa_index = 3

best_dsa_index = np.argmax(marks[:, dsa_index])

print("\n--- BEST DSA STUDENT ---")
print("Student :", students[best_dsa_index])
print("Marks   :", marks[best_dsa_index, dsa_index])


# ============================================================
# BONUS 2: Students with 90+ in Python
# ============================================================

python_index = 2

python_90 = marks[:, python_index] >= 90

print("\n--- STUDENTS WITH 90+ IN PYTHON ---")

for i in range(num_students):
    if python_90[i]:
        print(students[i])


# ============================================================
# BONUS 3: Rahul's Best Subject
# ============================================================

rahul_index = np.where(students == "Rahul")[0][0]

rahul_best_subject_index = np.argmax(marks[rahul_index])

print("\n--- RAHUL'S BEST SUBJECT ---")
print("Subject :", subjects[rahul_best_subject_index])
print("Marks   :", marks[rahul_index, rahul_best_subject_index])


# ============================================================
# BONUS 4: Students Below 60 in At Least One Subject
# ============================================================

below_60 = np.any(marks < 60, axis=1)

print("\n--- STUDENTS BELOW 60 IN AT LEAST ONE SUBJECT ---")

for i in range(num_students):
    if below_60[i]:
        print(students[i])


# ============================================================
# BONUS 5: Student Percentage
# ============================================================

percentage = np.mean(marks, axis=1)

print("\n--- STUDENT PERCENTAGE ---")

for i in range(num_students):
    print(f"{students[i]:8} -> {percentage[i]:.2f}%")
```
